```python

from roboflow import Roboflow
rf = Roboflow(api_key="YOUR_ROBOFLOW_API_KEY")  # Replace with your actual Roboflow API key
project = rf.workspace("alphajack2006-gmail-com").project("vietnamese-car-license-plate-vlbzz")
version = project.version(1)
dataset = version.download("yolov8")
                
```

```python
from ultralytics import YOLO
print("\n[1/3] Initializing YOLOv8 Nano base model on GPU...\n")
model = YOLO("yolov8n.pt") 
```

```python
print("[2/3] Fine-Tuning model on License Plate Dataset...")
# to get dataset via dataset.location/data.yaml

results = model.train(
    data=f"{dataset.location}/data.yaml", # Path to downloaded dataset.yaml
    epochs=16,                             # 16 epoch for first round testing
    imgsz=416,                             # lowered img resolution
    batch=32,                              # Batch size i increase to try and reduce training time and utilise more gpu
    name="vietnam_plate_lpr",              # Output folder
    save=True,                             # Save best.pt weights
    verbose=True,
    patience=2,
)
```
