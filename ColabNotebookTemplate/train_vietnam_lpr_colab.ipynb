# Step 3: Download Vietnamese License Plate Dataset from Roboflow

```python
from roboflow import Roboflow

# Enter your Roboflow API key below
rf = Roboflow(api_key="YOUR_ROBOFLOW_API_KEY")
project = rf.workspace("alphajack2006-gmail-com").project("vietnamese-car-license-plate-vlbzz")
version = project.version(1)
dataset = version.download("yolov8")
```

# Step 4: Train YOLOv8 Model with High Resolution and Full Convergence

```python
from ultralytics import YOLO

# Load pre-trained nano weights
model = YOLO("yolov8n.pt")

# Training with imgsz=640 and epochs=50 to capture small 2-line plates
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    patience=10,
    name="vietnam_plate_lpr",
    save=True
)
```
